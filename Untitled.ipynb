```python
# Block 1: Imports & Setup
# (We skip the install line because you did it in Terminal)

import pandas as pd
import os
from llama_cpp import Llama
from tqdm import tqdm 

# Verify paths
base_path = "trial data"
model_filename = "Qwen2.5-7B-Instruct-Q4_K_M.gguf" 
model_path = os.path.join(base_path, model_filename)

if os.path.exists(model_path):
    print(f"✅ Found model at: {model_path}")
else:
    print(f"❌ ERROR: Model not found at {model_path}. Check filename/folder.")
```

```python
# Load the model with specific settings for Mac Air (Metal/GPU)
llm = Llama(
    model_path=model_path,
    n_gpu_layers=-1,  # -1 = Offload EVERYTHING to Mac GPU
    n_ctx=4096,       # Context window (allows for long questions)
    verbose=False     # Keeps the output clean
)

print("✅ Model loaded successfully on Mac GPU!")
```

```python
# Load the datasets
mcq_path = os.path.join(base_path, "trial_data_multiple_choice.tsv")
saq_path = os.path.join(base_path, "trial_data_unique_answer.tsv")

df_mcq = pd.read_csv(mcq_path, sep='\t')
df_saq = pd.read_csv(saq_path, sep='\t')

print(f"📊 Data Loaded:\n- MCQ Track: {len(df_mcq)} questions\n- SAQ Track: {len(df_saq)} questions")
display(df_mcq.head(2))
```

```python
# --- STEP 1: AUTO-FIX DATA COLUMNS ---
def clean_columns(df, track_type):
    """
    Renames columns to standard formats (option_a, option_b, etc.)
    regardless of how they are named in the source file.
    """
    df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]
    
    # Check for common variations and map them to standard names
    # Example: 'option a' -> 'option_a', 'answer a' -> 'option_a'
    for letter in ['a', 'b', 'c', 'd']:
        # Try to find the column for this letter
        potential_cols = [c for c in df.columns if letter in c and ('option' in c or 'answer' in c or len(c) < 3)]
        if potential_cols:
            # Sort by length to pick the most likely candidate (usually the shortest or most descriptive)
            best_match = sorted(potential_cols, key=len)[-1] 
            df.rename(columns={best_match: f'option_{letter}'}, inplace=True)
            
    print(f"✅ Data Columns Fixed for {track_type}. Current columns: {list(df.columns)}")
    return df

# Apply the fix immediately
df_mcq = clean_columns(df_mcq, "MCQ")
df_saq = clean_columns(df_saq, "SAQ")

# --- STEP 2: UPDATED SOLVER FUNCTION ---
def solve_question(row, task_type="SAQ"):
    """
    Generates the answer using a culture-specific persona.
    """
    # 1. Extract context (Robust extraction)
    # We use .get() to avoid errors if a column is missing
    country = row.get('country', row.get('region', 'the target region'))
    language = row.get('language', 'English')
    question = row.get('question', '')
    
    # 2. Define the Cultural Persona
    system_prompt = (
        f"You are a knowledgeable local resident of {country}. "
        f"You speak {language} fluently and understand the deep cultural nuances, traditions, and everyday life of {country}. "
        "Answer as a native would."
    )

    # 3. Construct the Prompt
    if task_type == "MCQ":
        # Now uses the normalized column names
        options = f"A) {row.get('option_a', '')}\nB) {row.get('option_b', '')}\nC) {row.get('option_c', '')}\nD) {row.get('option_d', '')}"
        user_prompt = (
            f"Question: {question}\n\n{options}\n\n"
            f"Task: Select the option that is most culturally appropriate for {country}. "
            "Return ONLY the letter (A, B, C, or D) and nothing else."
        )
        max_tokens = 10
    else: # SAQ
        user_prompt = (
            f"Question: {question}\n\n"
            f"Task: Answer the question directly in {language}. "
            f"Your answer must reflect the specific cultural context of {country}. "
            "Be concise and accurate."
        )
        max_tokens = 200

    # 4. Run Inference
    try:
        output = llm.create_chat_completion(
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.1,
            max_tokens=max_tokens
        )
        response_text = output['choices'][0]['message']['content'].strip()
    except Exception as e:
        return "Error"

    # 5. Clean MCQ Answers
    if task_type == "MCQ":
        for char in response_text:
            if char.upper() in ['A', 'B', 'C', 'D']:
                return char.upper()
        return "A" 
        
    return response_text
```

```python
# 1. Run MCQ Track
print("🚀 Starting MCQ Track Inference...")
tqdm.pandas(desc="MCQ Progress")
df_mcq['prediction'] = df_mcq.progress_apply(lambda row: solve_question(row, task_type="MCQ"), axis=1)

# 2. Run SAQ Track
print("\n🚀 Starting SAQ Track Inference...")
tqdm.pandas(desc="SAQ Progress")
df_saq['prediction'] = df_saq.progress_apply(lambda row: solve_question(row, task_type="SAQ"), axis=1)

print("\n✅ Inference Complete!")
```

```python
import zipfile

# --- HELPER: FIND THE ID COLUMN ---
def get_submission_df(df, track_name):
    """
    Finds the ID column and the Prediction column to create the submission file.
    """
    # 1. Find the ID column
    # Look for columns that contain "id" (case insensitive) or just take the first column
    id_col = None
    for col in df.columns:
        if "id" in col.lower():
            id_col = col
            break
    
    # Fallback: If no column has "id" in the name, use the very first column
    if id_col is None:
        id_col = df.columns[0]
    
    print(f"✅ For {track_name}, using ID column: '{id_col}'")

    # 2. Create the submission dataframe
    # We strictly need the ID and the 'prediction' we generated earlier
    submission = df[[id_col, 'prediction']].copy()
    
    # 3. Rename columns to standard 'id' and 'prediction' just to be safe for CodaBench
    submission.columns = ['id', 'prediction']
    return submission

# --- EXECUTE SUBMISSION CREATION ---

# 1. Prepare dataframes
submission_mcq = get_submission_df(df_mcq, "MCQ")
submission_saq = get_submission_df(df_saq, "SAQ")

# 2. Save TSV files (Tab Separated Values)
submission_mcq.to_csv("submission_mcq.tsv", sep='\t', index=False)
submission_saq.to_csv("submission_saq.tsv", sep='\t', index=False)

# 3. Zip them together
zip_filename = "submission.zip"
with zipfile.ZipFile(zip_filename, 'w') as zipf:
    zipf.write("submission_mcq.tsv")
    zipf.write("submission_saq.tsv")

print(f"\n🎉 FINAL SUBMISSION READY: {zip_filename}")
print("You can now upload 'submission.zip' to CodaBench!")
```

```python
import pandas as pd
import os
import shutil
import zipfile
from tqdm import tqdm
from llama_cpp import Llama

# --- CONFIGURATION ---
# Define your model path explicitly (pointing to the trial data folder)
MODEL_PATH = "trial data/Qwen2.5-7B-Instruct-Q4_K_M.gguf"

# 1. ROBUST LANGUAGE MAPPING
# Maps codes (e.g., msa-SG) to the full cultural context.
CODE_TO_CONTEXT = {
    'msa-SG': ('Malay', 'Singapore'), 'tam-SG': ('Tamil', 'Singapore'), 'zho-SG': ('Chinese', 'Singapore'),
    'spa-EC': ('Spanish', 'Ecuador'), 'eng-GB': ('English', 'United Kingdom'), 'zho-CN': ('Chinese', 'China'),
    'spa-ES': ('Spanish', 'Spain'), 'spa-MX': ('Spanish', 'Mexico'), 'ind-ID': ('Indonesian', 'Indonesia'),
    'kor-KR': ('Korean', 'South Korea'), 'ell-GR': ('Greek', 'Greece'), 'fas-IR': ('Persian', 'Iran'),
    'ara-EG': ('Arabic', 'Egypt'), 'ara-MA': ('Arabic', 'Morocco'), 'ara-SA': ('Arabic', 'Saudi Arabia'),
    'eng-AU': ('English', 'Australia'), 'eus-ES': ('Basque', 'Spain'), 'fra-FR': ('French', 'France'),
    'gle-IE': ('Irish', 'Ireland'), 'tam-LK': ('Tamil', 'Sri Lanka'), 'tgl-PH': ('Tagalog', 'Philippines'),
    'bul-BG': ('Bulgarian', 'Bulgaria'), 'jpn-JP': ('Japanese', 'Japan'),
    # Fallbacks
    'eng-US': ('English', 'United States')
}

def get_context(code):
    """Returns (Language, Country) tuple from the code."""
    if not isinstance(code, str): return ('English', 'the target region')
    clean_code = code.split('_')[0] 
    return CODE_TO_CONTEXT.get(clean_code, ('English', 'the target region'))

# --- 2. LOAD MODEL ---
print(f"Loading Model from {MODEL_PATH}...")
if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(f"❌ Error: Model not found at {MODEL_PATH}. Make sure the file is in the 'trial data' folder.")

llm = Llama(
    model_path=MODEL_PATH,
    n_gpu_layers=-1, # Force Mac GPU
    n_ctx=4096,
    verbose=False
)
print("✅ Model Loaded.")

# --- 3. SMART PROMPTING FUNCTION ---
def solve_row(row, track_type):
    # A. Decode Context
    # Try multiple common column names for the ID/Region code
    code_source = row.get('lang_reg', row.get('id', 'eng-US'))
    lang, country = get_context(code_source)
    
    # Robust question extraction
    question = row.get('question', row.get('Question', ''))
    
    # B. Persona System Prompt
    system_prompt = (
        f"You are a culturally aware local resident of {country}. "
        f"You speak {lang} natively. Your goal is to explain the local customs, "
        f"traditions, and everyday knowledge of {country} accurately."
    )

    # C. Handle Formats
    options_text = ""
    
    if track_type == "MCQ":
        # Handle different column formats (Official vs Trial)
        if 'option_a' in row: # Official Data often uses lowercase with underscore
            opts = [row.get('option_a'), row.get('option_b'), row.get('option_c'), row.get('option_d')]
        elif 'Option A' in row: # Sometimes it has spaces/caps
            opts = [row.get('Option A'), row.get('Option B'), row.get('Option C'), row.get('Option D')]
        elif 'multiple_choice_options' in row: # Trial Data format
            opts = str(row['multiple_choice_options']).split('\n')
        else:
            opts = ["Option A", "Option B", "Option C", "Option D"]

        # Ensure we have 4 options
        while len(opts) < 4: opts.append("N/A")
        
        options_text = f"A) {opts[0]}\nB) {opts[1]}\nC) {opts[2]}\nD) {opts[3]}"
        
        user_prompt = (
            f"Question: {question}\n\n{options_text}\n\n"
            f"Task: Which option is the most culturally correct for {country}?\n"
            "Think step-by-step about the culture, then output strictly:\n"
            "ANSWER: [Letter]"
        )
        # Allow room for reasoning
        max_tokens = 150 
        
    else: # SAQ
        user_prompt = (
            f"Question: {question}\n\n"
            f"Task: Answer this question directly based on {country} culture in {lang}.\n"
            "Keep it concise."
        )
        max_tokens = 200

    # D. Inference
    try:
        output = llm.create_chat_completion(
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.1
        )
        response = output['choices'][0]['message']['content'].strip()
    except Exception as e:
        return "A" if track_type == "MCQ" else "Error"

    # E. Post-Processing
    if track_type == "MCQ":
        # Regex to find the LAST single letter answer given
        import re
        # Looks for patterns like "Answer: B" or just "B"
        matches = re.findall(r'\b([A-D])\b', response.upper())
        if matches:
            return matches[-1]
        return "A" # Fallback
    
    return response

# --- 4. DATA PROCESSING & SUBMISSION GENERATOR ---
def run_pipeline(mcq_file, saq_file):
    
    # Output Setup
    out_dir = "prediction"
    if os.path.exists(out_dir): shutil.rmtree(out_dir)
    os.makedirs(out_dir)

    # --- TRACK 2: MCQ ---
    print(f"\n🚀 Processing MCQ: {mcq_file}")
    if not os.path.exists(mcq_file):
        print(f"❌ File not found: {mcq_file}")
        return

    df_mcq = pd.read_csv(mcq_file, sep='\t')
    # CRITICAL FIX: Clean column names (remove spaces/case sensitivity issues)
    df_mcq.columns = [c.strip().lower().replace(" ", "_") for c in df_mcq.columns]
    
    # Run Inference
    tqdm.pandas(desc="MCQ Inference")
    df_mcq['final_pred'] = df_mcq.progress_apply(lambda r: solve_row(r, "MCQ"), axis=1)
    
    # Format for CodaBench (One-Hot Encoding)
    sub_mcq = pd.DataFrame()
    # Find ID column (could be 'id' or 'index')
    id_col = next((col for col in df_mcq.columns if 'id' in col or 'index' in col), df_mcq.columns[0])
    sub_mcq['id'] = df_mcq[id_col]
    
    for opt in ['A', 'B', 'C', 'D']:
        sub_mcq[opt] = (df_mcq['final_pred'] == opt).astype(int)
        
    sub_mcq.to_csv(f"{out_dir}/track_2_mcq_prediction.tsv", sep='\t', index=False)
    print("✅ MCQ File Generated.")

    # --- TRACK 1: SAQ ---
    print(f"\n🚀 Processing SAQ: {saq_file}")
    if not os.path.exists(saq_file):
        print(f"❌ File not found: {saq_file}")
        return

    df_saq = pd.read_csv(saq_file, sep='\t')
    # CRITICAL FIX: Clean column names
    df_saq.columns = [c.strip().lower().replace(" ", "_") for c in df_saq.columns]
    
    # Run Inference
    tqdm.pandas(desc="SAQ Inference")
    df_saq['final_pred'] = df_saq.progress_apply(lambda r: solve_row(r, "SAQ"), axis=1)
    
    # Format for CodaBench
    sub_saq = pd.DataFrame()
    id_col = next((col for col in df_saq.columns if 'id' in col or 'index' in col), df_saq.columns[0])
    sub_saq['id'] = df_saq[id_col]
    sub_saq['prediction'] = df_saq['final_pred']
    
    sub_saq.to_csv(f"{out_dir}/track_1_saq_prediction.tsv", sep='\t', index=False)
    print("✅ SAQ File Generated.")

    # --- ZIP ---
    shutil.make_archive("submission", 'zip', root_dir='.', base_dir='prediction')
    print("\n🎉 DONE. Upload 'submission.zip' to CodaBench.")

# --- 5. EXECUTE ---
# FIXED: Pointing to the "trial data" folder where your files actually are
run_pipeline(
    mcq_file="trial data/trial_data_multiple_choice.tsv", 
    saq_file="trial data/trial_data_unique_answer.tsv"
)
```
